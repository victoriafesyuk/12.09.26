```python
import gradio as gr

def process_text(text):

    if text.endswith("!"):
        sentence = "Окличне речення"
    elif text.endswith("?"):
        sentence = "Питальне речення"
    elif text.endswith("."):
        sentence = "Розповідне речення"
    else:
        sentence = "Неможливо визначити"

    return sentence

interface = gr.Interface(
fn=process_text,
inputs="text",
outputs="text"
)

interface.launch()
```

```python
import gradio as gr
import re


def top5_words(text):

    words = re.findall(r'\b\w+\b', text.lower())
    if not words:
        return "Слів не знайдено"

    freq = {}

    for word in words:
        freq[word] = freq.get(word, 0) + 1

    def get_count(x):
        return x[1]

    sorted_words = sorted(
    freq.items(),
    key=get_count,
    reverse=True
    )[:5]

    top5 = [f"\n{i+1}. {word}"
                   for i, (word, _) in enumerate(sorted_words)]

    return "\n".join(top5)

def word_count(text):
    words = re.findall(r'\b\w+\b', text)
    return len(words)

def sentence_count(text):
    sentences = re.split(r'[.!?]+', text)
    sentences = [s.strip() for s in sentences if s.strip()]

    return len(sentences)

def avg_sentence(text):
    sentences = re.split(r'[.!?]+', text)
    sentences = [s.strip() for s in sentences if s.strip()]

    words = re.findall(r'\b\w+\b', text)

    if not sentences:
        return 0

    return f"{round(len(words) / len(sentences), 2)} слів"

def avg_word(text):

    words = re.findall(r'\b\w+\b', text)
    if not words:
        return 0
    total_length = sum(len(word) for word in words)
    return round(total_length / len(words), 2)

def unique_words(text):
    words = re.findall(r'\b\w+\b', text.lower())
    unique = sorted(set(words))
    return "\n".join(unique)

def text_stats(text):
    text_top5 = top5_words(text)
    text_unique = unique_words(text)
    text_word_count = word_count(text)
    text_sentence_count = sentence_count(text)
    text_avg_word = avg_word(text)
    text_avg_sentence = avg_sentence(text)

    return (
        f"### Лексичний аналіз\n\n"
        f"Топ 5 слів:{text_top5}\n\n"
        f"Унікальні слова: {text_unique}\n\n"
        f"### Підрахунки\n\n"
        f"Кількість слів: {text_word_count}\n\n"
        f"Кількість речень: {text_sentence_count}\n\n"
        f"Середня довжина слова: {text_avg_word}\n\n"
        f"Середня довжина речення:{text_avg_sentence}"
    )

demo = gr.Interface(
    fn=text_stats,
    inputs=gr.Textbox(label="Ваш текст", placeholder="Введіть текст"),
    outputs=gr.Markdown(),
    title="Аналіз введеного тексту",
    submit_btn="Проаналізувати",
    clear_btn="Очистити",
    flagging_mode="never"

)
demo.launch()
```
